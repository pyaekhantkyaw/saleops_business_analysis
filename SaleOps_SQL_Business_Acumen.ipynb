# Product Performance

<h2 id="q1">1.1 Question 1: Revenue Contribution & Percentage Share by Product Line</h2>

```python
%%sql
CREATE TEMP VIEW ranked_products AS  
WITH product_summary AS (
    SELECT 
        warehouse,
        product_line, 
        SUM(quantity) AS total_quantity, 
        ROUND(SUM(total * (1 - payment_fee)), 2) AS net_revenue
    FROM sales 
    GROUP BY warehouse, product_line
)
SELECT 
    warehouse,
    ROW_NUMBER() OVER (PARTITION BY warehouse ORDER BY net_revenue DESC) AS line_rank,
    product_line, 
    total_quantity, 
    net_revenue,
    SUM(net_revenue) OVER (PARTITION BY warehouse) AS total_warehouse_revenue, 
    ROUND(100.0 * net_revenue / SUM(net_revenue) OVER (PARTITION BY warehouse), 1) AS contribution
FROM product_summary;
```

```python
%%sql
WITH product_summary AS (
    SELECT 
        warehouse,
        product_line, 
        SUM(quantity) AS total_quantity, 
        ROUND(SUM(total * (1 - payment_fee)), 2) AS net_revenue
    FROM sales 
    GROUP BY warehouse, product_line
)
SELECT 
    warehouse,
    ROW_NUMBER() OVER (PARTITION BY warehouse ORDER BY net_revenue DESC) AS line_rank,
    product_line, 
    total_quantity, 
    net_revenue,
    SUM(net_revenue) OVER (PARTITION BY warehouse) AS total_warehouse_revenue, 
    ROUND(100.0 * net_revenue / SUM(net_revenue) OVER (PARTITION BY warehouse), 1) AS contribution
FROM product_summary;
```

Displaying the ranking of product lines within each warehouse

```python
%%sql
SELECT * 
FROM ranked_products 
ORDER BY warehouse, line_rank;
```

<h4><b>Key Insight:</b> </h4>
<b>Suspension & Traction</b> and <b>Frame & Body</b> consistently deliver the highest revenue share across warehouses (~20–30% each), making them the <b>primary drivers of overall sales performance.</b> Other lines contribute smaller shares, indicating secondary importance for revenue growth.

<a id="q2"></a>
<h3>1.2 Question 2: Product Line Prioritization for Revenue Maximization</h3>

```python
%%sql
SELECT 
    warehouse,
    product_line AS dominant_product_line,
    net_revenue,
    contribution
FROM ranked_products
WHERE line_rank = 1;
```

<h4><b>Key Insights:</b></h4> Prioritize Suspension & Traction as the <b>primary income stream</b>, while expanding focus on Frame & Body for its strong balance of demand and pricing power, which form the core drivers of profit growth. 

----

#### (i) Additional Finding: Which product line contributes the least revenue? 
Engine is the lowest‑revenue contributor across warehouses, with Miscellaneous also underperforming in Central.

```python
%%sql
WITH weakest AS ( 
    SELECT 
        product_line, 
        warehouse, 
        ROUND(SUM(total * (1 - payment_fee)), 2) AS net_revenue
    FROM sales 
    GROUP BY product_line, warehouse) ,
ranked AS (
    SELECT *,
        DENSE_RANK() OVER (
            PARTITION BY Warehouse 
            ORDER BY net_revenue ASC
        ) AS rank
    FROM weakest 
) 
SELECT * 
FROM ranked
WHERE rank = 1 
```

# Warehouse Performance

<a id="q3"></a>
<h3>2.1 Question 3: Key Drivers of Revenue Differences Between Warehouses</h3>

```python
%%sql
SELECT 
    warehouse, 
    SUM(quantity) AS total_quantity,
    ROUND(SUM(total * (1 - payment_fee)), 2) AS net_revenue,
    ROUND(AVG(unit_price), 2) AS avg_selling_price 
FROM sales 
GROUP BY warehouse
ORDER BY net_revenue DESC;
```

<h4><b>Key Insight:</b></h4>
The primary revenue difference is highly influenced by the place, showing the <b>Location specific demand </b>.It explains why the central outperfroms than the others. In addition, the price differences contribute only marginally to overall revenue disaprities indicating the demand and sales volume and the dominant factors.

<h4><b>Key Insight: </b></h4>

---

#### (ii) Additional Finding: Top two product lines per warehouse
Suspension & Traction and Frame & Body are the universal revenue anchors across all warehouses, forming the strategic backbone of sales performance.

```python
%%sql
SELECT 
    Product_Line,
    Warehouse,
    net_revenue,
    total_quantity
FROM (
    SELECT 
        Product_Line, 
        Warehouse, 
        ROUND(SUM(total * (1 - payment_fee)), 2) AS net_revenue, 
        SUM(Quantity) AS total_quantity,
        DENSE_RANK() OVER (
            PARTITION BY Warehouse 
            ORDER BY ROUND(SUM(total * (1 - payment_fee)), 2) DESC
        ) AS rank
    FROM sales
    GROUP BY Product_Line, Warehouse
) t
WHERE rank <= 2 ;
```

# Customer Analysis

<a id="q5"></a>
<h3>3.1 Question 5: Client Type Influence on Purchasing Behavior & Profitability</h3>

```python
%%sql 
SELECT 
    Client_Type, 
    warehouse, 
    ROUND(SUM(total * (1 - payment_fee)), 2) AS net_revenue,
    SUM(quantity) AS total_quantity ,
    ROUND ( AVG(Unit_Price),2 ) AS avg_unit_price 
FROM sales 
GROUP BY Client_Type, warehouse 
```

<h4><b>Key Insight:</b></h4>
Wholesale operations in the <b>Central and North</b > through bulk purchase, while <b>Retail Clients in the West </b>generate higher revenue per unit which highlights a location‑specific pricing or demand effect.
<h4>1.Purchasing Behavior</h4> 
Wholesale clients buy in bulk, driving large volumes in Central and North.
Retail clients in the West purchase fewer units but at higher per‑unit prices.
</br> 
<h4>2.Market Dynamics </h4>  Profitability depends on location.  Central and North benefit from wholesale bulk margins, while West shows stronger retail unit pricing <b> local demand effect.</b>
<h4>3.Strategy Implication </h4>
If a company wants to buy in bulk, West warehouse should be considered, making best area to source. Reselling those goods in Central where demand and sales volumes are the highest, would maximize the profit potential.

<h3>3.2 Question 6: Revenue Per Order by Client Type</h3>

```python
%%sql 
SELECT 
    Client_Type, 
    ROUND(AVG(net_revenue), 2 ) AS avg_revenue_per_order
FROM (
    SELECT 
        client_type, 
        order_number, 
        SUM(total * (1 - payment_fee)) AS net_revenue
    FROM sales 
    GROUP BY client_type, order_number 
)t 
GROUP BY client_type 
ORDER BY avg_revenue_per_order DESC ;
```

<b>Finding:</b> Wholesales clients consistnetly generate higher revenue per order, reflecting their bulk-buying nature and lower transaction fees.(Bank Transfer ~1%) 

<b>Complementary Insight:</b> This findings complements the previous finding, where wholesale buyers dominated profitablity in Central and North as of the volume it sold, while Retail client in the West benefited from higher per-unit prices despite smaller orders. 

<h4><b>Key Insight:</b></h4> 
Profitability is shaped by client type: </br>
Wholesale → scales through order size and lower fees. </br>
Retail → depends on pricing power and local demand conditions.

# Sales Trend Analysis

<a id="q7"></a>
<h3>4.1 Question 7: Daily Revenue Over Time (Daily Total)</h3>

```python
%%sql
SELECT 
    DATE(date) as date,
    ROUND(SUM(total), 1) AS daily_total  
FROM sales
GROUP BY DATE(date)
ORDER BY date
LIMIT 10 ;
```

------

<a id="q8"></a>
<h3>4.2 Question 8: Revenue Stability vs. Short-Term Fluctuations (7-Day Moving Average)</h3>

```python
%%sql
WITH daily_revenue AS (
    SELECT 
        DATE(date) AS date,          
        ROUND(SUM(total), 1) AS daily_total   
    FROM sales
    GROUP BY DATE(date)
)
SELECT 
    date,
    ROUND(daily_total, 2) AS daily_total,
    ROUND(
        AVG(daily_total) OVER (
            ORDER BY date 
            ROWS BETWEEN 6 PRECEDING AND CURRENT ROW
        ), 2
    ) AS rolling_7day_avg,
    ROUND(
        LAG(daily_total, 1) OVER (ORDER BY date), 2
    ) AS previous_day_revenue
FROM daily_revenue
ORDER BY date
LIMIT 10 ;
```

<h4><b>Key Insight:</b></h4>
Revenue Stability Trend: Reveal a stable band around ~2,600–3,000.  </br>
Revenue Volatility: Daily revenue shows significant short‑term fluctuations (under 1k to over 5k within a week).
In short, daily revenue fluctuates sharply, but the 7-Day moving average shows stable performance around 2.6-3.0k, which indicate resilience against short-term swings.

<h2><a id="q9"></a>4.3 Question 9: Day-over-Day Revenue Growth & Decline (Growth Percentage)</h2>

```python
%%sql
WITH daily_revenue AS (
    SELECT 
        DATE(date) AS date,              -- strip time, keep only YYYY-MM-DD
        ROUND(SUM(total), 1) AS daily_total   -- round daily total
    FROM sales
    GROUP BY DATE(date)
)
SELECT 
    date,
    ROUND(daily_total, 2) AS daily_total,   
    ROUND(
        AVG(daily_total) OVER (
            ORDER BY date 
            ROWS BETWEEN 6 PRECEDING AND CURRENT ROW
        ), 2
    ) AS rolling_7day_avg,
    ROUND(
        LAG(daily_total, 1) OVER (ORDER BY date), 2
    ) AS previous_day_revenue,
    ROUND(
        (daily_total - LAG(daily_total, 1) OVER (ORDER BY date)) 
        / NULLIF(LAG(daily_total, 1) OVER (ORDER BY date), 0) * 100, 
        2
    ) AS day_over_day_growth_pct
FROM daily_revenue
ORDER BY date
LIMIT 10 ;
```

<h4><b>Key Insight:</b></h4>
Revenue shows sharp dayily swings (-75% to +220%)
